```python
import pandas as pd

import numpy as np
import torch
with open("../../parameters/device_params_modified.txt") as f:
    content = f.read()

device_params = {
    k: torch.tensor([float(v)])
for v, k in [i.split(" # ") for i in content.splitlines()]}

from networker_pytorch import NanoOptoElectronicNeuron
```

```python
neuron = NanoOptoElectronicNeuron(
    N=1,
    unity_coefficient=None,
    **device_params
)
```

```python
mem = neuron.init_neuron()


fc1 = torch.nn.Linear(1, 2)
# Record the final layer
spk1_rec = []
mem1_rec = []
dt = 0.0001
x = torch.tensor([1.0])

num_steps = 100
for step in range(num_steps):
    inh_curr, exc_curr = fc1(x)
    spk1, mem1 = neuron.forward(x, 0, mem, dt)
    
    spk1_rec.append(spk1)
    mem1_rec.append(mem1)
```
